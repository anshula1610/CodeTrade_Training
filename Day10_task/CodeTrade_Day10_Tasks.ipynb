***Task 1***

```python
import pandas as pd
customers = pd.DataFrame({

    "customer_id": [1, 2, 3, 4],

    "customer_name": [
        "Anuj",
        "Rahul",
        "Priya",
        "Neha"
    ],

    "region": [
        "North",
        "South",
        "East",
        "West"
    ],

    "segment": [
        "Consumer",
        "Corporate",
        "Consumer",
        "Home Office"
    ]
})

products = pd.DataFrame({

    "product_id": [101, 102, 103],

    "product_name": [
        "Laptop",
        "Phone",
        "Tablet"
    ],

    "category": [
        "Electronics",
        "Electronics",
        "Electronics"
    ],

    "price": [
        50000,
        25000,
        30000
    ]
})

orders = pd.DataFrame({

    "order_id": [1001, 1002, 1003, 1004],

    "customer_id": [1, 2, 3, 4],

    "product_id": [101, 102, 103, 101],

    "quantity": [1, 2, 1, 3]
})

def audit_table(df, table_name):

    print(f"\n--- {table_name} ---")

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns)

    print("\nData Types:")
    print(df.dtypes)

    print("\nNull Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nUnique Counts:")
    print(df.nunique())

audit_table(customers, "Customers Table")

audit_table(products, "Products Table")

audit_table(orders, "Orders Table")
```

***Task 3***

```python
import pandas as pd

sales = pd.DataFrame({
    "region":["North","South","North","East","West"],
    "category":["Electronics","Furniture","Furniture","Electronics","Electronics"],
    "segment":["Consumer","Corporate","Consumer","Home Office","Corporate"],
    "sales":[50000,30000,20000,40000,35000]
})

region_sales = sales.groupby("region")["sales"].sum()
print(region_sales)

category_sales = sales.groupby("category")["sales"].sum()
print(category_sales)

segment_sales = sales.groupby("segment")["sales"].sum()
print(segment_sales)

multi_group = sales.groupby(["region","segment"])["sales"].sum()
print(multi_group)
```

***Task 4***

```python
import pandas as pd

customers = pd.DataFrame({
    "customer_id":[1,2,3],
    "customer_name":["Anuj","Rahul","Priya"]
})

products = pd.DataFrame({
    "product_id":[101,102],
    "product_name":["Laptop","Phone"],
    "price":[50000,25000]
})

orders = pd.DataFrame({
    "order_id":[1001,1002,1003],
    "customer_id":[1,2,3],
    "product_id":[101,102,101],
    "quantity":[1,2,1]
})

merged = pd.merge(orders,customers,on="customer_id")
merged = pd.merge(merged,products,on="product_id")

merged["revenue"] = merged["price"] * merged["quantity"]

total_revenue = merged["revenue"].sum()
print("Total Revenue:",total_revenue)

average_order_value = merged["revenue"].mean()
print("Average Order Value:",average_order_value)

top_products = merged.groupby("product_name")["quantity"].sum().sort_values(ascending=False)
print(top_products)

print(merged)
```

```python
import pandas as pd

sales = pd.DataFrame({
    "region":["North","South","North","East","West"],
    "month":["Jan","Jan","Feb","Feb","Jan"],
    "category":["Electronics","Furniture","Furniture","Electronics","Electronics"],
    "segment":["Consumer","Corporate","Consumer","Home Office","Corporate"],
    "sales":[50000,30000,20000,40000,35000]
})

pivot1 = sales.pivot_table(
    values="sales",
    index="region",
    columns="month",
    aggfunc="sum",
    fill_value=0
)

print(pivot1)

pivot2 = sales.pivot_table(
    values="sales",
    index="category",
    columns="segment",
    aggfunc="sum",
    fill_value=0
)

print(pivot2)
```

***Task 8***

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("sales.db")

sales = pd.DataFrame({
    "customer_name":["Anuj","Rahul","Priya","Neha"],
    "region":["North","South","East","West"],
    "sales":[50000,30000,40000,35000]
})

sales.to_sql(
    "sales_table",
    conn,
    if_exists="replace",
    index=False
)

query1 = "SELECT * FROM sales_table;"
print(pd.read_sql(query1, conn))

query2 = """
SELECT customer_name,sales
FROM sales_table
WHERE sales > 35000;
"""
print(pd.read_sql(query2, conn))

query3 = """
SELECT region,SUM(sales) AS total_sales
FROM sales_table
GROUP BY region;
"""
print(pd.read_sql(query3, conn))

query4 = """
SELECT *
FROM sales_table
ORDER BY sales DESC;
"""
print(pd.read_sql(query4, conn))

query5 = """
SELECT AVG(sales) AS average_sales
FROM sales_table;
"""
print(pd.read_sql(query5, conn))

conn.close()
```

***Task 9***

```python
import pandas as pd
import sqlite3

sales = pd.DataFrame({
    "region":["North","South","North","East","West"],
    "sales":[50000,30000,20000,40000,35000]
})

pandas_result = sales.groupby("region")["sales"].sum()

print("Pandas Result:")
print(pandas_result)

conn = sqlite3.connect("sales.db")

sales.to_sql(
    "sales_table",
    conn,
    if_exists="replace",
    index=False
)

sql_query = """
SELECT region,
SUM(sales) AS total_sales
FROM sales_table
GROUP BY region;
"""

sql_result = pd.read_sql(sql_query, conn)

print("\nSQL Result:")
print(sql_result)

conn.close()

print("\nComparison:")
print("Pandas is more flexible for data analysis within Python.")
print("SQL is better for querying data stored in databases.")
print("Both produce the same business result but use different syntax.")
```

Pandas and SQL both produced the same sales summary by region.

Pandas is more flexible because it integrates directly with Python and supports advanced data manipulation.

SQL is more suitable when data is stored inside databases and needs to be queried efficiently.

For small analyses, Pandas is often easier to read. For large database systems, SQL is generally preferred.
