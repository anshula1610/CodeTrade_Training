```python
import google.generativeai as genai
import pandas as pd

genai.configure(api_key="Enter your API key")

models = [
    "models/gemini-2.0-flash",
    "models/gemini-2.5-flash",
    "models/gemini-2.5-pro"
]

prompt = "Explain Artificial Intelligence in 100 words."

results = []

for model_name in models:
    try:
        model = genai.GenerativeModel(model_name)
        response = model.generate_content(prompt)

        results.append({
            "Model": model_name,
            "Response Length": len(response.text),
            "Response": response.text[:300]
        })

        print(f"✓ Completed: {model_name}")

    except Exception as e:
        print(f"✗ Error in {model_name}: {e}")

df = pd.DataFrame(results)

df
```

```python
df.to_csv("llm_comparison_results.csv", index=False)
print("Results saved successfully!")
```

```python
import google.generativeai as genai
import pandas as pd

genai.configure(api_key="Enter your API key")

model = genai.GenerativeModel("models/gemini-2.5-flash")

prompts = {
    "Zero Shot": "Explain Machine Learning.",
    "Few Shot": """
Example:
AI = Artificial Intelligence
ML = Machine Learning
DL = Deep Learning

Explain Data Science in the same style.
""",
    "Chain of Thought": """
Explain step by step how a student can learn Data Science from scratch.
"""
}

results = []

for prompt_type, prompt in prompts.items():
    try:
        response = model.generate_content(prompt)

        results.append({
            "Prompt Type": prompt_type,
            "Response": response.text
        })

        print(f"✓ {prompt_type} completed")

    except Exception as e:
        print(f"✗ Error in {prompt_type}: {e}")

df = pd.DataFrame(results)
df
```

***Task 3: Streaming AI Chat Assistant***

```python
import google.generativeai as genai
genai.configure(api_key="Enter your API key")
model = genai.GenerativeModel("models/gemini-2.5-flash")
```

```python
while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        print("Chat ended!")
        break

    response = model.generate_content(user_input)

    print("\nAI:")
    print(response.text)
    print("-" * 50)
```

***Report***

User can ask multiple questions in a single session.
The assistant generates real-time responses.
Exit command terminates the conversation

***Task 4: Token Usage & Cost Tracker***

```python
import google.generativeai as genai
import pandas as pd

genai.configure(api_key="Enter the API key")

model = genai.GenerativeModel("models/gemini-2.5-flash")

prompt = "Explain Artificial Intelligence."

response = model.generate_content(prompt)

response_text = response.text

# Approximate token count
token_count = len(response_text.split())

# Approximate cost calculation
cost_per_1000_tokens = 0.0001

estimated_cost = (token_count / 1000) * cost_per_1000_tokens

results = pd.DataFrame({
    "Prompt": [prompt],
    "Tokens Used": [token_count],
    "Estimated Cost ($)": [estimated_cost]
})

print(results)
```

```python
results.to_csv("token_usage_report.csv", index=False)
print("Token usage report saved!")
```

***Report Points for Task 4***
Tracked approximate token usage.
Estimated API cost based on token count.
Stored results in a structured DataFrame.
Exported token statistics to CSV for analysis.
